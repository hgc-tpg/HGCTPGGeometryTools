```python
import matplotlib.pyplot as plt
import numpy as np
from shapely.plotting import plot_polygon
from geometrychecking import data, geometry, plotting
```

```python
geometry_name = "test_triggergeom"
```

```python
stcs = geometry.stcs_from_json(f'../resources/{geometry_name}/geojson')
```

```python
fig, axes = plt.subplots(nrows=7, ncols=3, figsize=(9, 21), dpi=150)
for layer,layer_stcs in stcs.items():
    ax = axes[(layer-1)//3, (layer-1)%3]
    for stc in layer_stcs:
        stcid = stc['stcid']
        stc_poly = stc['shape']
        buffer = stc_poly.length/1000
        plot_polygon(stc_poly.buffer(buffer).buffer(-buffer), ax=ax, color='lightgrey', add_points=False, linewidth=0.5)
        #plot_polygon(stc_poly, ax=ax, color='lightgrey', add_points=False, linewidth=0.5)
    ax.set_title(f"Layer {layer}")
plt.tight_layout()
plt.show()
```
